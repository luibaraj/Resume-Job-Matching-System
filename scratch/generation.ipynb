```python
import ollama

MODEL = "llama3.2:3b-instruct-q4_K_M"

# Verify Ollama server is up and model is available
models = ollama.list()
model_names = [m.model for m in models.models]
print("Available models:", model_names)
assert MODEL in model_names, f"Model '{MODEL}' not found. Run: ollama pull {MODEL}"
print(f"✓ Model ready: {MODEL}")
```

```python
def generate(prompt: str, model: str = MODEL) -> str:
    response = ollama.chat(
        model=model,
        messages=[{"role": "user", "content": prompt}],
        options={
            "temperature": 0.7,
            "top_p": 0.9,
            "num_predict": 150,  # matches original max_length=150
        },
    )
    return response["message"]["content"]

print("✓ Generation function ready")
```

```python
prompt = "What are the key skills for a software engineer?"
response = generate(prompt)
print(f"Prompt: {prompt}\n\nResponse:\n{response}")
```
